```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df['Cabin'] = df['Cabin'].fillna("Unknown")
```

```python
duplicates_before = df.duplicated().sum()

print(duplicates_before)
```

```python
df.drop_duplicates(inplace=True)
```

```python
duplicates_after = df.duplicated().sum()

print(duplicates_after)
```

```python
df['Sex'] = df['Sex'].replace({
    'male':'Male',
    'Male':'Male',
    'M':'Male',

    'female':'Female',
    'Female':'Female',
    'F':'Female'
})
```

```python
df['Sex'] = df['Sex'].replace({
    'male': 'Male',
    'female': 'Female'
})
```

```python
print(df.columns)
```

```python
Q1 = df['Age'].quantile(0.25)

Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1
```

```python
upper = Q3 + 1.5*IQR
```

```python
lower = Q1 - 1.5*IQR
```

```python
outliers = df[(df['Age']<lower) | (df['Age']>upper)]

outliers
```

```python
df = df[(df['Age']>=lower) & (df['Age']<=upper)]
```

```python
df = df.copy()

df.loc[:, 'Age'] = np.where(df['Age'] > upper, upper, df['Age'])
df.loc[:, 'Age'] = np.where(df['Age'] < lower, lower, df['Age'])
```

```python
df.dtypes
```

```python
df['PassengerId'] = df['PassengerId'].astype(str)
```

```python
df['Fare'] = df['Fare'].astype(float)
```

```python
# Save original statistics BEFORE cleaning
original_rows = len(df)
original_nulls = df.isnull().sum().sum()   # Total null values
original_duplicates = df.duplicated().sum()

print(original_rows)
print(original_nulls)
print(original_duplicates)
```

```python
summary = pd.DataFrame({
    'Before': [original_rows, original_nulls, original_duplicates],
    'After': [
        len(df),
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
}, index=['Rows', 'Null Values', 'Duplicates'])

summary
```
